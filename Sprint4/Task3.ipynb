Создать базовый класс Subscription в C#, который будет представлять подписки на

Требования к базовому классу Subscription:

• Атрибуты: ID подписки (SubscriptionId), Название услуги (ServiceName), 
Стоимость подписки (Cost).

• Методы:
1. CalculateMonthlyCost(): метод для расчета ежемесячной стоимости 
подписки.
2. ExtendSubscription(): метод для продления подписки на 
дополнительный период.
3. GetSubscriptionDetails(): метод для получения деталей подписки.

Требования к производным классам:
1. ПодпискаНаОнлайнСервис (OnlineServiceSubscription): Должна содержать 
дополнительные атрибуты, такие как Количество доступных пользователей 
(MaxUsers). Метод CalculateMonthlyCost() должен быть переопределен для 
учета количества пользователей при расчете стоимости.
2. ПодпискаНаСтreamинг (StreamingSubscription): Должна содержать 
дополнительные атрибуты, такие как Количество одновременных потоков 
(MaxStreams). Метод ExtendSubscription() должен быть переопределен для 
добавления специальных предложений для продления подписки.
3. ПодпискаНаВидео(VideoSubscription) (если требуется третий класс): Должна 
содержать дополнительные атрибуты, такие как Качество видео 
(VideoQuality). Метод GetSubscriptionDetails() должен быть переопределен 
для отображения качества видео вместе с другими деталями подписки.

```
public interface IDiscountable
{
    double GetDiscount();
}

public interface IAutoRenewable
{
    void EnableAutoRenewal();
}

public class Subscription
{
    public static int Count { get; private set; } = 0;

    private int _subscriptionId;
    private string _serviceName;
    private double _cost;
    private int _durationMonths;

    public int SubscriptionId
    {
        get { return _subscriptionId; }
        set { _subscriptionId = value; }
    }

    public string ServiceName
    {
        get { return _serviceName; }
        set { _serviceName = value; }
    }

    public double Cost
    {
        get { return _cost; }
        set
        {
            if (value >= 0)
                _cost = value;
            else
                throw new ArgumentException("Стоимость не может быть отрицательной.");
        }
    }

    public int DurationMonths
    {
        get { return _durationMonths; }
        set
        {
            if (value > 0)
                _durationMonths = value;
            else
                throw new ArgumentException("Срок подписки должен быть больше нуля.");
        }
    }

    public Subscription(int id, string serviceName, double cost, int durationMonths = 1)
    {
        SubscriptionId = id;
        ServiceName = serviceName;
        Cost = cost;
        DurationMonths = durationMonths;
        Count++;
    }

    public virtual double CalculateMonthlyCost()
    {
        return Cost;
    }

    public virtual void ExtendSubscription()
    {
        DurationMonths++;
        Console.WriteLine($"Подписка «{ServiceName}» продлена на месяц. Срок: {DurationMonths} мес.");
    }

    public void ExtendSubscription(int months)
    {
        if (months <= 0)
            throw new ArgumentException("Количество месяцев должно быть больше нуля.");

        DurationMonths += months;
        Console.WriteLine($"Подписка «{ServiceName}» продлена на {months} мес. Новый срок: {DurationMonths} мес.");
    }

    public virtual void GetSubscriptionDetails()
    {
        Console.WriteLine(
            $"ID: {SubscriptionId}, услуга: {ServiceName}, " +
            $"стоимость: {Cost:F2} руб./мес., срок: {DurationMonths} мес.");
    }

    public void CompareCost(Subscription other)
    {
        if (other == null)
            throw new ArgumentNullException(nameof(other));

        double thisCost = CalculateMonthlyCost();
        double otherCost = other.CalculateMonthlyCost();

        if (thisCost > otherCost)
            Console.WriteLine($"«{ServiceName}» дороже, чем «{other.ServiceName}».");
        else if (thisCost < otherCost)
            Console.WriteLine($"«{ServiceName}» дешевле, чем «{other.ServiceName}».");
        else
            Console.WriteLine($"«{ServiceName}» и «{other.ServiceName}» стоят одинаково.");
    }

    public static void ShowTotalSubscriptions()
    {
        Console.WriteLine($"Всего создано подписок: {Count}");
    }
}

public class OnlineServiceSubscription : Subscription, IDiscountable, IAutoRenewable
{
    private int _maxUsers;

    public int MaxUsers
    {
        get { return _maxUsers; }
        set
        {
            if (value > 0) _maxUsers = value;
            else throw new ArgumentException("Количество пользователей должно быть больше нуля.");
        }
    }

    public OnlineServiceSubscription(int id, string serviceName, double cost, int maxUsers, int durationMonths = 1)
        : base(id, serviceName, cost, durationMonths)
    {
        MaxUsers = maxUsers;
    }

    public override double CalculateMonthlyCost()
    {
        return Cost * MaxUsers;
    }

    double IDiscountable.GetDiscount()
    {
        return MaxUsers >= 5 ? 0.10 : 0.0;
    }

    void IAutoRenewable.EnableAutoRenewal()
    {
        Console.WriteLine($"Для онлайн-подписки «{ServiceName}» включено автопродление.");
    }

    public override void GetSubscriptionDetails()
    {
        Console.WriteLine(
            $"Онлайн-сервис: {ServiceName}, ID: {SubscriptionId}, пользователей: {MaxUsers}, " +
            $"стоимость за пользователя: {Cost:F2} руб., общая стоимость: {CalculateMonthlyCost():F2} руб./мес.");
    }
}

public class StreamingSubscription : Subscription, IDiscountable, IAutoRenewable
{
    private int _maxStreams;

    public int MaxStreams
    {
        get { return _maxStreams; }
        set
        {
            if (value > 0) _maxStreams = value;
            else throw new ArgumentException("Количество потоков должно быть больше нуля.");
        }
    }

    public StreamingSubscription(int id, string serviceName, double cost, int maxStreams, int durationMonths = 1)
        : base(id, serviceName, cost, durationMonths)
    {
        MaxStreams = maxStreams;
    }

    public override void ExtendSubscription()
    {
        DurationMonths++;
        Console.WriteLine(
            $"Стриминговая подписка «{ServiceName}» продлена на месяц. " +
            "Специальное предложение: скидка 5% на следующий месяц.");
    }

    double IDiscountable.GetDiscount()
    {
        return DurationMonths >= 6 ? 0.10 : 0.05;
    }

    void IAutoRenewable.EnableAutoRenewal()
    {
        Console.WriteLine($"Для стриминговой подписки «{ServiceName}» включено автопродление.");
    }

    public override void GetSubscriptionDetails()
    {
        Console.WriteLine(
            $"Стриминг: {ServiceName}, ID: {SubscriptionId}, потоков: {MaxStreams}, " +
            $"стоимость: {Cost:F2} руб./мес., срок: {DurationMonths} мес.");
    }
}

public class VideoSubscription : StreamingSubscription, IDiscountable, IAutoRenewable
{
    private string _videoQuality;

    public string VideoQuality
    {
        get { return _videoQuality; }
        set
        {
            if (value == "HD" || value == "Full HD" || value == "4K")
                _videoQuality = value;
            else
                throw new ArgumentException("Доступные качества: HD, Full HD или 4K.");
        }
    }

    public VideoSubscription(
        int id, string serviceName, double cost, int maxStreams,
        string videoQuality, int durationMonths = 1)
        : base(id, serviceName, cost, maxStreams, durationMonths)
    {
        VideoQuality = videoQuality;
    }

    public override void GetSubscriptionDetails()
    {
        Console.WriteLine(
            $"Видеоподписка: {ServiceName}, ID: {SubscriptionId}, качество: {VideoQuality}, " +
            $"потоков: {MaxStreams}, стоимость: {Cost:F2} руб./мес., срок: {DurationMonths} мес.");
    }

    double IDiscountable.GetDiscount()
    {
        return VideoQuality == "4K" ? 0.15 : 0.10;
    }

    void IAutoRenewable.EnableAutoRenewal()
    {
        Console.WriteLine($"Для видеоподписки «{ServiceName}» включено автопродление.");
    }
}

public interface ISubscriptionLogger
{
    void Log(string message);
}

public class ConsoleSubscriptionLogger : ISubscriptionLogger
{
    public void Log(string message)
    {
        Console.WriteLine(message);
    }
}

public class SubscriptionManager
{
    private List<Subscription> _subscriptions = new List<Subscription>();
    private readonly ISubscriptionLogger _logger;

    public SubscriptionManager(ISubscriptionLogger logger)
    {
        _logger = logger;
    }

    public void AddSubscription(Subscription subscription)
    {
        if (subscription == null)
            throw new ArgumentNullException(nameof(subscription));

        _subscriptions.Add(subscription);
        _logger.Log($"Добавлена подписка: {subscription.ServiceName}");
    }

    public void ShowAllSubscriptions()
    {
        Console.WriteLine("\n=== Все подписки ===");
        foreach (Subscription subscription in _subscriptions)
        {
            subscription.GetSubscriptionDetails();
        }
    }

    public void ShowMonthlyTotal()
    {
        double total = 0;
        foreach (Subscription subscription in _subscriptions)
            total += subscription.CalculateMonthlyCost();

        Console.WriteLine($"\nОбщая стоимость всех подписок за месяц: {total:F2} руб.");
    }

    public void CompareAllWith(Subscription subscription)
    {
        foreach (Subscription item in _subscriptions)
        {
            if (item.SubscriptionId != subscription.SubscriptionId)
                item.CompareCost(subscription);
        }
    }
}

public class SubscriptionStorage<T> where T : Subscription
{
    private List<T> _items = new List<T>();

    public void Add(T subscription)
    {
        _items.Add(subscription);
        Console.WriteLine($"В хранилище добавлена подписка: {subscription.ServiceName}");
    }

    public void ShowAll()
    {
        Console.WriteLine("\n=== Подписки в Generic-хранилище ===");
        foreach (T subscription in _items)
        {
            subscription.GetSubscriptionDetails();
        }
    }
}

var online = new OnlineServiceSubscription(1, "Рабочка по подпискам", 150, 4, 3);
var streaming = new StreamingSubscription(2, "Музыка и кино", 500, 3, 6);
var video = new VideoSubscription(3, "Премиум-видео", 700, 4, "4K", 12);

var logger = new ConsoleSubscriptionLogger();
var manager = new SubscriptionManager(logger);
manager.AddSubscription(online);
manager.AddSubscription(streaming);
manager.AddSubscription(video);

manager.ShowAllSubscriptions();
manager.ShowMonthlyTotal();

Console.WriteLine("\n=== Продление ===");
online.ExtendSubscription();
streaming.ExtendSubscription();
video.ExtendSubscription();

Console.WriteLine("\n=== Сравнение стоимости ===");
online.CompareCost(streaming);
streaming.CompareCost(video);
manager.CompareAllWith(video);

Console.WriteLine("\n=== Дополнительные возможности ===");
((IAutoRenewable)online).EnableAutoRenewal();
((IAutoRenewable)streaming).EnableAutoRenewal();
((IAutoRenewable)video).EnableAutoRenewal();

Console.WriteLine("\n=== Скидки через интерфейс ===");
Console.WriteLine($"Скидка онлайн-подписки: {((IDiscountable)online).GetDiscount() * 100}%");
Console.WriteLine($"Скидка стриминговой подписки: {((IDiscountable)streaming).GetDiscount() * 100}%");
Console.WriteLine($"Скидка видеоподписки: {((IDiscountable)video).GetDiscount() * 100}%");

Console.WriteLine("\n=== Итог ===");
Subscription.ShowTotalSubscriptions();

Console.WriteLine("\n=== Generic-класс ===");
var storage = new SubscriptionStorage<Subscription>();
storage.Add(online);
storage.Add(streaming);
storage.Add(video);
storage.ShowAll();

Console.WriteLine("\n=== Перегрузка метода ===");
online.ExtendSubscription(2);
```
