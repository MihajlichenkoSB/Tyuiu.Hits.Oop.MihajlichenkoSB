```polyglot-notebook
using System;
using System.Collections.Generic;


public class Product
{
    private string _name;
    private decimal _price;
    private int _quantity;

    protected string category;

    internal string description;

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public decimal Price
    {
        get { return _price; }
        set
        {
            if (value >= 0)
            {
                _price = value;
            }
        }
    }

    public int Quantity
    {
        get { return _quantity; }
        set
        {
            if (value >= 0)
            {
                _quantity = value;
            }
        }
    }

    public string Category
    {
        get { return category; }
        set { category = value; }
    }

    public string Description
    {
        get { return description; }
        set { description = value; }
    }

    public Product(
        string name,
        decimal price,
        int quantity,
        string category,
        string description)
    {
        Name = name;
        Price = price;
        Quantity = quantity;
        Category = category;
        Description = description;
    }

    public decimal GetTotalPrice()
    {
        return Price * Quantity;
    }

    public void UpdateQuantity(int amount)
    {
        Quantity += amount;

        if (Quantity < 0)
        {
            Quantity = 0;
        }

        Console.WriteLine(
            $"Количество товара {Name}: {Quantity} шт."
        );
    }

    public string GetProductInfo()
    {
        return $"Название: {Name}\n" +
               $"Цена: {Price} руб.\n" +
               $"Количество: {Quantity} шт.\n" +
               $"Категория: {Category}\n" +
               $"Описание: {Description}";
    }
}

public class Seller
{
    // private поля
    private string _name;
    private string _employeeId;

    // protected поле
    protected decimal salary;

    // internal поле
    internal string contactInfo;

    // список товаров
    private List<Product> products;

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public string EmployeeId
    {
        get { return _employeeId; }
        set { _employeeId = value; }
    }

    public decimal Salary
    {
        get { return salary; }
        set { salary = value; }
    }

    public string ContactInfo
    {
        get { return contactInfo; }
        set { contactInfo = value; }
    }

    public Seller(
        string name,
        string employeeId,
        decimal salary,
        string contactInfo)
    {
        Name = name;
        EmployeeId = employeeId;
        Salary = salary;
        ContactInfo = contactInfo;

        products = new List<Product>();
    }

    public void AddProduct(Product product)
    {
        products.Add(product);

        Console.WriteLine(
            $"Продавец {Name} добавил товар: {product.Name}"
        );
    }

    public void SellProduct(Product product, int quantity)
    {
        if (quantity <= 0)
        {
            Console.WriteLine(
                "Количество товара должно быть больше 0."
            );

            return;
        }

        if (quantity > product.Quantity)
        {
            Console.WriteLine(
                $"Недостаточно товара {product.Name} на складе."
            );

            return;
        }

        product.UpdateQuantity(-quantity);

        Console.WriteLine(
            $"Продавец {Name} продал " +
            $"{quantity} шт. товара {product.Name}."
        );
    }

    public string GetSellerInfo()
    {
        return $"Имя: {Name}\n" +
               $"ID сотрудника: {EmployeeId}\n" +
               $"Зарплата: {Salary} руб.\n" +
               $"Контакты: {ContactInfo}";
    }


    public List<Product> GetProducts()
    {
        return products;
    }
}

public class Store
{
    private string _storeName;
    private string _location;

    public string storeHours;

    private List<Seller> sellers;

    public string StoreName
    {
        get { return _storeName; }
        set { _storeName = value; }
    }

    public string Location
    {
        get { return _location; }
        set { _location = value; }
    }

    public string StoreHours
    {
        get { return storeHours; }
        set { storeHours = value; }
    }

    public Store(
        string storeName,
        string location,
        string storeHours)
    {
        StoreName = storeName;
        Location = location;
        StoreHours = storeHours;

        sellers = new List<Seller>();
    }

    public void AddSeller(Seller seller)
    {
        sellers.Add(seller);

        Console.WriteLine(
            $"Продавец {seller.Name} добавлен в магазин."
        );
    }

    public void ListProducts()
    {
        Console.WriteLine();
        Console.WriteLine("=== ТОВАРЫ МАГАЗИНА ===");

        foreach (Seller seller in sellers)
        {
            Console.WriteLine();
            Console.WriteLine(
                $"Продавец: {seller.Name}"
            );

            foreach (Product product in seller.GetProducts())
            {
                Console.WriteLine(
                    $"- {product.Name} | " +
                    $"{product.Price} руб. | " +
                    $"{product.Quantity} шт."
                );
            }
        }
    }

    public string GetStoreInfo()
    {
        return $"Название: {StoreName}\n" +
               $"Часы работы: {StoreHours}\n" +
               $"Адрес: {Location}";
    }
}


Product laptop = new Product(
    "Ноутбук ASUS",
    85000m,
    5,
    "Ноутбуки",
    "Игровой ноутбук ASUS"
);

Product phone = new Product(
    "iPhone 15",
    70000m,
    8,
    "Смартфоны",
    "Смартфон Apple"
);

Product headphones = new Product(
    "AirPods Pro",
    25000m,
    12,
    "Наушники",
    "Беспроводные наушники"
);

Seller seller = new Seller(
    "Алексей",
    "EMP001",
    65000m,
    "alexYxtaPyxta@super.com"
);

Store store = new Store(
    "Техно Маркет",
    "ул. Линейная, д. 152, к. 3",
    "09:00-20:00"
);

Console.WriteLine("=== МАГАЗИН ===");
Console.WriteLine();

Console.WriteLine(store.GetStoreInfo());

Console.WriteLine();
Console.WriteLine("=== ПРОДАВЕЦ ===");
Console.WriteLine();

Console.WriteLine(seller.GetSellerInfo());

Console.WriteLine();

store.AddSeller(seller);


Console.WriteLine();
Console.WriteLine("=== ДОБАВЛЕНИЕ ТОВАРОВ ===");
Console.WriteLine();

seller.AddProduct(laptop);
seller.AddProduct(phone);
seller.AddProduct(headphones);

Console.WriteLine();
Console.WriteLine("=== ТОВАРЫ ===");
Console.WriteLine();

Console.WriteLine(laptop.GetProductInfo());

Console.WriteLine();

Console.WriteLine(phone.GetProductInfo());

Console.WriteLine();

Console.WriteLine(headphones.GetProductInfo());

Console.WriteLine();
Console.WriteLine("=== СТОИМОСТЬ ===");
Console.WriteLine();

Console.WriteLine(
    $"Стоимость всех ноутбуков: " +
    $"{laptop.GetTotalPrice()} руб."
);

Console.WriteLine(
    $"Стоимость всех телефонов: " +
    $"{phone.GetTotalPrice()} руб."
);

Console.WriteLine(
    $"Стоимость всех наушников: " +
    $"{headphones.GetTotalPrice()} руб."
);

Console.WriteLine();
Console.WriteLine("=== ПРОДАЖА ===");
Console.WriteLine();

seller.SellProduct(laptop, 2);

Console.WriteLine();

seller.SellProduct(phone, 1);


Console.WriteLine();
Console.WriteLine("=== ТОВАРЫ ПОСЛЕ ПРОДАЖИ ===");
Console.WriteLine();

Console.WriteLine(
    $"Ноутбук ASUS: {laptop.Quantity} шт."
);

Console.WriteLine(
    $"iPhone 15: {phone.Quantity} шт."
);

Console.WriteLine(
    $"AirPods Pro: {headphones.Quantity} шт."
);


store.ListProducts();
```
