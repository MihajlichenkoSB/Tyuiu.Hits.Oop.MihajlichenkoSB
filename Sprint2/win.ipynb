```polyglot-notebook
using System;
using System.Collections.Generic;

public class Employee
{
    public string Name { get; set; }
    public string Position { get; set; }
    public double Salary { get; set; }

    public Employee(string name, string position, double salary)
    {
        Name = name;
        Position = position;
        Salary = salary;
    }

    public virtual void PrintInfo()
    {
        Console.WriteLine($"Имя: {Name}");
        Console.WriteLine($"Должность: {Position}");
        Console.WriteLine($"Зарплата: {Salary:F2}");
    }
}

public class Manager : Employee
{
    public string Department { get; set; }

    public Manager(string name, string position, double salary, string department)
        : base(name, position, salary)
    {
        Department = department;
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Отдел: {Department}");
    }
}

public interface IReportable
{
    string GenerateReport();
}

public interface ITeamLeader
{
    string[] GetTeamMembers();
}

public class ProjectManager : Manager, IReportable, ITeamLeader
{
    public string ProjectName { get; set; }
    public string[] TeamMembers { get; set; }

    public ProjectManager(
        string name,
        string position,
        double salary,
        string department,
        string projectName,
        string[] teamMembers)
        : base(name, position, salary, department)
    {
        ProjectName = projectName;
        TeamMembers = teamMembers;
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Проект: {ProjectName}");
        Console.WriteLine($"Члены команды: {string.Join(", ", TeamMembers)}");
    }

    public string GenerateReport()
    {
        return $"Отчет по проекту «{ProjectName}»\n" +
               $"Ответственный: {Name}\n" +
               $"Отдел: {Department}\n" +
               $"Количество участников: {TeamMembers.Length}";
    }

    public string[] GetTeamMembers()
    {
        return TeamMembers;
    }
}

// Тестирование
Employee employee = new Employee("Иван Иванов", "Специалист", 65000);
Manager manager = new Manager("Мария Смирнова", "Менеджер", 85000, "Продажи");

ProjectManager projectManager = new ProjectManager(
    "Алексей Петров",
    "Проектный менеджер",
    110000,
    "Разработка",
    "Корпоративный портал",
    new string[] { "Анна", "Дмитрий", "Ольга" });

Console.WriteLine("=== Employee ===");
employee.PrintInfo();

Console.WriteLine("\n=== Manager ===");
manager.PrintInfo();

Console.WriteLine("\n=== ProjectManager ===");
projectManager.PrintInfo();

Console.WriteLine("\n=== Отчет ===");
Console.WriteLine(projectManager.GenerateReport());

Console.WriteLine("\n=== Участники команды ===");
foreach (string member in projectManager.GetTeamMembers())
{
    Console.WriteLine(member);
}
```
