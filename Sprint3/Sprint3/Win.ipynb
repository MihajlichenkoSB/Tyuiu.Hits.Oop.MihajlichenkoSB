```polyglot-notebook
public interface IInstallable
{
    void Install();
}

public abstract class Component
{
    public string Name { get; set; }
    public double Price { get; set; }

    public Component(string name, double price)
    {
        Name = name;
        Price = price;
    }

    public virtual void GetInfo()
    {
        Console.WriteLine($"{Name} — {Price} руб.");
    }

    public override string ToString()
    {
        return $"{Name} — {Price} руб.";
    }

    public static bool Same(Component a, Component b)
    {
        return a != null && b != null &&
               a.Name == b.Name && a.Price == b.Price;
    }
}

public class CPU : Component, IInstallable
{
    public CPU(string name, double price) : base(name, price) { }
    public override void GetInfo()
    {
        Console.WriteLine($"Процессор: {Name}, {Price} руб.");
    }
    public void Install() => Console.WriteLine($"Установили процессор {Name}");
    public static bool operator ==(CPU a, CPU b) => Same(a, b);
    public static bool operator !=(CPU a, CPU b) => !Same(a, b);
    public override bool Equals(object obj) => obj is CPU other && Same(this, other);
    public override int GetHashCode() => HashCode.Combine(Name, Price);
}

public class RAM : Component, IInstallable
{
    public RAM(string name, double price) : base(name, price) { }
    public override void GetInfo()
    {
        Console.WriteLine($"Память: {Name}, {Price} руб.");
    }
    public void Install() => Console.WriteLine($"Установили память {Name}");
    public static bool operator ==(RAM a, RAM b) => Same(a, b);
    public static bool operator !=(RAM a, RAM b) => !Same(a, b);
    public override bool Equals(object obj) => obj is RAM other && Same(this, other);
    public override int GetHashCode() => HashCode.Combine(Name, Price);
}

public class GPU : Component, IInstallable
{
    public GPU(string name, double price) : base(name, price) { }
    public override void GetInfo()
    {
        Console.WriteLine($"Видеокарта: {Name}, {Price} руб.");
    }
    public void Install() => Console.WriteLine($"Установили видеокарту {Name}");
    public static bool operator ==(GPU a, GPU b) => Same(a, b);
    public static bool operator !=(GPU a, GPU b) => !Same(a, b);
    public override bool Equals(object obj) => obj is GPU other && Same(this, other);
    public override int GetHashCode() => HashCode.Combine(Name, Price);
}

public class HDD : Component, IInstallable
{
    public HDD(string name, double price) : base(name, price) { }
    public override void GetInfo()
    {
        Console.WriteLine($"Диск: {Name}, {Price} руб.");
    }
    public void Install() => Console.WriteLine($"Установили диск {Name}");
    public static bool operator ==(HDD a, HDD b) => Same(a, b);
    public static bool operator !=(HDD a, HDD b) => !Same(a, b);
    public override bool Equals(object obj) => obj is HDD other && Same(this, other);
    public override int GetHashCode() => HashCode.Combine(Name, Price);
}

public class Computer
{
    private List<Component> parts = new List<Component>();

    public void AddComponent<T>(T part) where T : Component, IInstallable
    {
        part.Install();
        parts.Add(part);
    }

    public void RemoveComponent<T>(T part) where T : Component
    {
        if (parts.Remove(part))
            Console.WriteLine($"Удалили: {part.Name}");
        else
            Console.WriteLine("Такой детали нет.");
    }

    public double GetTotalPrice()
    {
        double total = 0;
        foreach (Component part in parts)
            total += part.Price;
        return total;
    }

    public void ShowParts()
    {
        foreach (Component part in parts)
            part.GetInfo();
    }
}

public static class ComputerFactory
{
    public static Computer Create<T1, T2, T3, T4>(T1 cpu, T2 ram, T3 gpu, T4 hdd)
        where T1 : Component, IInstallable
        where T2 : Component, IInstallable
        where T3 : Component, IInstallable
        where T4 : Component, IInstallable
    {
        Computer pc = new Computer();
        pc.AddComponent(cpu);
        pc.AddComponent(ram);
        pc.AddComponent(gpu);
        pc.AddComponent(hdd);
        return pc;
    }
}

CPU cpu = new CPU("Intel i5", 20000);
RAM ram = new RAM("16 GB", 5000);
GPU gpu = new GPU("RTX 4060", 35000);
HDD hdd = new HDD("1 TB", 4000);

Computer pc = ComputerFactory.Create(cpu, ram, gpu, hdd);

Console.WriteLine("\nДетали компьютера:");
pc.ShowParts();

Console.WriteLine($"\nОбщая цена: {pc.GetTotalPrice()} руб.");

CPU cpu2 = new CPU("Intel i5", 20000);
Console.WriteLine($"Процессоры одинаковые: {cpu == cpu2}");

pc.RemoveComponent(hdd);
Console.WriteLine($"Цена после удаления диска: {pc.GetTotalPrice()} руб.");
```
